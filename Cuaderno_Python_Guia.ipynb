### Detalle fino: `input()` siempre devuelve texto

Este es el error número uno de quien empieza. `input()` **no** entiende números:

```python
edad = input("¿Qué edad tienes? ")   # el usuario escribe: 30
print(edad + 1)   # ERROR: can only concatenate str to str
```

La solución es convertir el texto al momento de leerlo:

```python
edad = int(input("¿Qué edad tienes? "))    # ahora sí es un número entero
print(edad + 1)                            # 31
```

En el archivo del día la edad se guarda como texto y por eso se imprime
directamente con la f-string. En el [día 02](#d002) ya se empieza a usar
`int()` y `float()`.

- Cada `input()` **detiene** el programa hasta que el usuario pulse ENTER.
- `\n` dentro de un `print` produce un salto de línea; `print()` a secas también

## Día 02 — Tipos de datos, conversión y operadores

**Archivo:** `d002.py`

---


### Qué se ve en este archivo

Python es de **tipado dinámico**: no declaras el tipo de una variable, pero todo
valor tiene un tipo. Los cuatro tipos básicos:

| Tipo | Ejemplo | Para qué sirve |
| --- | --- | --- |
| `int` | `7` | Números enteros |
| `float` | `769.97` | Números con decimales |
| `str` | `'texto'` | Texto (siempre entre comillas) |
| `bool` | `True` / `False` | Verdad o falsedad |

Para preguntar el tipo de un valor: `type(valor)` → devuelve `<class 'int'>`.

### Conversión de tipos

Es la operación más usada del día. Se llama **casting** y consiste en envolver el
valor:

```python
texto = "42"
numero = int(texto)          # '42' -> 42
decimal = float("3.14")      # '3.14' -> 3.14
palabra = str(99)            # 99 -> '99'
```

Sin conversión, `'7' + '3'` da `'73'` (pega textos) mientras que `7 + 3` da `10`
(suma números).

### Operadores aritméticos

```text
a + b     suma                 a ** b    potencia
a - b     resta                a // b    división entera (sin decimales)
a * b     multiplicación       a %  b    resto (módulo) de la división
a / b     división (siempre float)
```

Nota sobre la línea comentada del archivo: *"el operador `//` devuelve resultado
de una división solo con número entero"*. Es correcto: `7 // 2` → `3` mientras
que `7 / 2` → `3.5`. Y `7 % 2` → `1`, que es el resto.

### El programa del día

Calculadora de exponentes: pide dos números con `int(input(...))`, aplica `**` y
muestra el resultado con f-string. Fíjate en que el `print` lleva `\n` para dejar
aire en la consola.

````python
# TIPOS DE DATOS
# Los string los divide en caracteres pudiendo acceder a cualquier posición

#El operador // devuelve resultado de una división solo con numero entero

# numero_1 = 4
# numero_2 = 769.97
# texto = 'am I a string'
# decision = True
# print(f"El valor {numero_2} es bastante más grande que {numero_1}. ¿{texto}? The answer is {decision}")

"""
Calculadora de exponentes, deberá pedirle al usuario los números que quiere.

"""
print('--- Calculadora de exponentes --- \n')
numero_1=int(input("Introduzca el primer número"))
numero_2=int(input("Introduzca el segundo número"))
resultado = numero_1 ** numero_2
print(f"El resultado de {numero_1} elevado a {numero_2} es {resultado}")
````

### Potencia: el operador `**`

```python
print(2 ** 3)      # 8
print(2 ** 10)     # 1024
print(9 ** 0.5)    # 3.0  -> raíz cuadrada
print(2 ** -1)     # 0.5  -> exponente negativo = división
```

El orden de prioridad de los operadores (de mayor a menor):

```text
**   ->  *, /, //, %   ->  +, -   ->  paréntesis
```

Los paréntesis mandan: `2 + 3 * 4` es `14`, pero `(2 + 3) * 4` es `20`.

### Errores típicos

| Error | Explicación |
| --- | --- |
| `'2' ** 3` | No se puede elevar un texto. Usa `int('2') ** 3` |
| `int('3.5')` | Falla: hay un punto. Usa `int(float('3.5'))` o directamente `float('3.5')` |
| `5 / 0` | `ZeroDivisionError`. Comprueba siempre el divisor antes |
| Olvidar `float` y perder decimales | `7 / 2` con `int` da `3.0`, no `3` |

### Idea clave del día

`input()` da `str`. Si el programa va a **operar** con el dato, conviértelo
inmediato con `int()` o `float()`: es la diferencia entre que el programa
funcione o reviente.

### Qué se ve en este archivo

Este día tiene tres bloques comentados de teoría y el proyecto final. La teoría

### 1. Condicionales: `if` / `elif` / `else`

Permiten que el programa **decida** qué camino seguir según una condición.

```python
if numero > 7:
    print("Es mayor que 7")
elif numero == 7:
    print("Es igual a 7")
else:
    print("Es menor que 7")
```

- Se evalúan de arriba abajo.
- En cuanto una condición es cierta, se ejecuta su bloque y **se salta el resto**.
- El `else` es opcional y siempre va el último.

**La indentación no es decorativa: es la sintaxis de Python.** Lo que está
sangrado 4 espacios pertenece al `if`. Un `IndentationError` es el error más
frecuente al escribir estos bloques.

### 2. Los operadores de comparación

```text
>  mayor que              >=  mayor o igual que
<  menor que              <=  menor o igual que
== igual que              !=  distinto de
```

Aviso: `=` **asigna**, `==` **compara**. Es el error de sintaxis lógica más común.

### 3. Condicionales anidados y operadores lógicos

Se pueden meter `if` dentro de `if` (el ejemplo del alcohol usa un menú de
bebidas dentro del `if` de la edad), y se combinan condiciones con:

```text
and   ->  las dos cosas deben ser ciertas
or    ->  basta con que una sea cierta
not   ->  invierte el valor
```

```python
if color == "rojo" and forma == "circulo":
    print("Es un círculo rojo")
```

### 4. `match` — el equivalente a `switch` de Java

```python
match error:
    case "200": print("Todo ok.")
    case "404": print("Página no encontrada.")
    case _:     print("Error no disponible.")
```

- Se compara el valor de `match` con cada `case` **en orden**.
- `case _` es el equivalente a `default`: se ejecuta si nada coincide.
- Cada `case` puede tener sus propios `if` anidados dentro.

## Día 04 — Listas y tuplas

**Archivo:** `d004.py`

---


### Qué se ve en este archivo

### 1. Listas

Una **lista** es una variable que guarda varios valores, **de distintos tipos**,
en orden, y a la que se accede **por posición** (índice).

```python
lista_colores = ["rojo", "azul", "amarillo", "verde"]
```

Los índices empiezan en **0**. El último es `-1`, el penúltimo `-2`.

```text
índice:     0        1          2         3
elemento: rojo    azul     amarillo   verde
```

### 2. Los métodos que se usan en el archivo

| Método | Qué hace | Ejemplo del archivo |
| --- | --- | --- |
| `lista[i] = x` | Sustituye un valor por su posición | `lista_colores[1] = 'naranja'` |
| `lista[i]` | Lee un valor | `print(lista_colores[0])` |
| `.append(x)` | Añade al final | `lista_colores.append('azul')` |
| `.insert(pos, x)` | Añade en una posición concreta | `lista_colores.insert(2, "violeta")` |
| `.pop(i)` | Elimina **por posición** y lo devuelve | `lista_colores.pop(0)` |
| `.remove(x)` | Elimina **por valor** | `lista_colores.remove('violeta')` |
| `.copy()` | Copia el contenido en otra lista | `lista2 = lista_colores.copy()` |
| `.count(x)` | Cuántas veces aparece un valor | `print(lista2.count('amarillo'))` |
| `.index(x)` | En qué posición está un valor | `print(lista2.index('naranja'))` |
| `.reverse()` | Da la vuelta a la lista | `lista2.reverse()` |
| `.sort()` | Ordena ascendente (alfabético o numérico) | `lista2.sort()` |
| `.sort(reverse=True)` | Ordena descendente | `lista2.sort(reverse=True)` |
| `.extend(otra)` | Añade todos los elementos de otra lista | `lista_colores.extend(lista2)` |
| `.clear()` | Vacía la lista | (comentado en el archivo) |
| `len(lista)` | Número de elementos | `print(len(lista_colores))` |

### 3. Truco: `lista[3][0]`

```python
print(lista_colores[3][0])   # imprime 'v'
```

Se lee de dentro hacia fuera: `lista_colores[3]` es `'verde'` (un texto) y
`'verde'[0]` es su primer carácter, `'v'`. Sirve para cuando una lista contiene
listas o textos. Si el elemento fuera un número, `lista[3][0]` daría error.

### 4. Tuplas

```python
mi_tupla = (2, 4, 6, 8, 10)
```

Son listas **inmutables**: se pueden leer y recorrer, pero **no se pueden
modificar**. No existen `append`, `remove`, `sort`...

```python
mi_tupla[1] = 20      # TypeError: 'tuple' object does not support item assignment
```

¿Para qué sirven entonces? Porque son **más seguras**: si un dato no debe
cambiar, se declara como tupla. También son más ligeras y se pueden usar como
claves de diccionario.

### `pop` frente a `remove`: no se confundan

```python
lista = ["rojo", "azul", "verde"]
lista.pop(0)          # elimina por POSICIÓN: el 'rojo'  -> ["azul", "verde"]
lista.remove("verde") # elimina por VALOR                    -> ["azul"]
```

Si el valor no existe, `remove` da `ValueError`. Si la posición está fuera de
rango, `pop` da `IndexError`.

### `list.copy()` frente a `list()`

```python
a = [1, 2, 3]
b = a.copy()     # copia INDEPENDIENTE
c = a            # MISMA lista (dos nombres, un objeto)
c.append(4)      # a también vale [1, 2, 3, 4]
```

Asignar una lista a otra variable **no** la copia. Es un fallo muy común cuando
crees que estás trabajando sobre una copia y en realidad estás modificando el
original. `.copy()` (o `a[:]`) sí crea una copia de verdad.

```python
matriz = [[1, 2, 3],

```

### `enumerate` y `zip` (introducidos en el [día 05](#d005))

listas a la vez:

```python
colores = ['rojo', 'azul']

for c, p in zip(colores, precios):

```

Un **diccionario** guarda pares **clave : valor**. La clave es la "etiqueta" que

```python
office = {
```

A diferencia de las listas, **el orden no importa**: se accede por nombre, no por

```python
office['outlook'] = 'Correo electrónico'    # si la clave NO existe, la crea
```

## Día 08 — Tkinter: la primera ventana

**Archivo:** `d008.py`

---


### Qué se ve en este archivo

Tkinter es la biblioteca de **interfaz gráfica de la biblioteca estándar**: no
hay que instalar nada, viene con Python. Sirve para hacer programas de
escritorio con botones, campos de texto y menús.

### El esqueleto de cualquier programa con Tkinter

```python
from tkinter import *    # 1. Importar el módulo

root = Tk()              # 2. Crear la ventana principal (el "root")

mensaje = Label(root, text='Mi primer programa con Tkinter')   # 3. Crear widgets
mensaje.pack()           # 4. Colocarlos en la ventana

root.mainloop()          # 5. Mantener la ventana abierta
```

Paso a paso:

1. **`from tkinter import *`** — importa todo lo público del módulo. La forma
   alternativa y más recomendable en proyectos grandes es
   `import tkinter as tk` y usar `tk.Label(...)`.
2. **`root = Tk()`** — el objeto raíz: la ventana. Se le puede poner el título
   (`root.title(...)`) y el tamaño (`root.geometry(...)`).
3. **`Label(root, text=...)`** — un widget. **Siempre** recibe `root` como primer
   argumento: le dice a qué ventana pertenece. Se puede guardar en una variable
   (`mensaje = ...`) o no (`Label(...).pack()`); guardarlo es necesario si luego
   hay que modificarlo.
4. **`pack()`** — coloca el widget dentro de la ventana. Sin llamar a un método de
   colocación (`pack`, `grid` o `place`), el widget **no aparece**. Es el error
   más común al empezar.
5. **`root.mainloop()`** — el bucle de eventos: Tkinter se queda esperando a que
   el usuario haga clic, escriba o cierre. **El programa no termina hasta que se
   cierra la ventana.**

### Ejecutar

Este archivo **no** se ejecuta desde el cuaderno: `mainloop()` abriría una
ventana y bloquearía el kernel. En una terminal:

```powershell
python d008.py
```

````python
# importaciones
from tkinter import * # Del módulo tkinter importalo todo

# Creación de la ventana principal
root = Tk()

# Creación de la etiqueta
mensaje = Label(root, text='Mi primer programa con Tkinter')

# Muestro la etiqueta
mensaje.pack()
# Bucle para mantener la ventana en ejecución
root.mainloop()
````

### `pack()` frente a `grid()` y `place()`

Tkinter tiene tres sistemas de colocación y **no se pueden mezclar** en la misma
ventana:

| Método | Cómo coloca | Útil para |
| --- | --- | --- |
| `pack(side=...)` | Apila los widgets en un orden fijo | Disposiciones simples, en fila o columna |
| `grid(row=, column=)` | Tabla de filas y columnas | Formularios alineados (el [día 09](#d009)) |
| `place(x=, y=)` | Coordenadas exactas en píxeles | Control total, poco usado |

Con `pack()` el orden de las llamadas **es** el orden de colocación, y no se
puede usar `side` en dos widgets que se solapen.

### `mainloop()` y el bucle de eventos

`mainloop()` es el equivalente en Tkinter del bucle `while` del
[día 05](#d005): un bucle infinito que espera eventos (clics, teclado, cerrar
ventana) y ejecuta la función asociada. Por eso el flujo de un programa gráfico
no es lineal como el de uno de consola, sino **reactivo a eventos**.

### Estructura mental de un programa gráfico

```text
1. Configurar la ventana        root.title(), root.geometry()
2. Crear los widgets            Label, Entry, Button, ...
3. Colocarlos en la ventana     .pack(), .grid(), .place()
4. Definir qué pasa al actuar   def funcion(): ...   (o lambdas)
5. Esperar                     root.mainloop()
```

Los puntos 4 y 5 se ven en detalle en el [día 12](#d012), que es donde aparecen
los botones, los eventos y el `command`.

## Día 09 — Tkinter: `title`, `geometry` y `grid`

**Archivo:** `d009.py`

---


### Qué se ve en este archivo

Se amplía la ventana del [día 08](#d008) con configuración y con un segundo
sistema de colocación.

### 1. Título de la ventana

```python
root.title('Curso de Tkinter')
```

Aparece en la barra superior de la ventana.

### 2. Tamaño y posición

```python
root.geometry('800x600+560+20')
```

El formato es `ancho x alto + desplazamiento_X + desplazamiento_Y` (en píxeles):

```text
800x600+560+20
 │    │  └── desplazamiento en X: 560 píxeles desde el borde izquierdo
 │    └───── desplazamiento en Y: 20 píxeles desde arriba
 └────────── alto: 600 píxeles
```

El desplazamiento es **opcional**: `root.geometry('800x600')` fija solo el tamaño y
deja que el sistema operativo coloque la ventana donde quiera. En la del archivo,
`+560+20` la sitúa a 560 píxeles del borde izquierdo y a 20 del borde superior.

### 3. Varias etiquetas y el método `grid`

```python
mensaje  = Label(root, text='Mi primer programa con Tkinter.')
mensaje2 = Label(root, text='Esta es la segunda etiqueta.')

mensaje.grid(row=0, column=0)
mensaje2.grid(row=1, column=1)
```

`grid()` coloca cada widget en una **tabla** de filas y columnas. Es el método
adecuado para formularios, porque dos widgets de la misma columna quedan
alineados.

El comentario del archivo apunta una forma de abreviar:

```python
# Se puede aplicar el grid directamente en la creación, ahorrando líneas
Label(root, text='Mi primer programa con Tkinter.').grid(row=0, column=0)
```

Ambas formas son equivalentes. La versión corta se usa cuando **no hace falta**
guardar la referencia al widget; si después hay que cambiar su texto o color, hay
que guardarlo en una variable.

Opciones útiles de `grid`:

```python
mensaje.grid(row=0, column=0, padx=10, pady=5, sticky='w')
#                  margen      margen    alineación al oeste (izquierda)
```

- `padx` / `pady`: espacio en píxeles alrededor del widget.
- `sticky`: `'w'`, `'e'`, `'n'`, `'s'` y combinaciones, para anclar el widget
  dentro de su celda y que se estire con la ventana.

`grid` y `pack` **no se pueden combinar** en la misma ventana:Tkinter lanza
`TclError` al usar los dos.

````python
# importaciones
from tkinter import * # Del módulo tkinter importalo todo

# Creación de la ventana principal
root = Tk()

# Título de la ventana
root.title('Curso de Tkinter')

# Tamaño de la ventana
root.geometry('800x600+560+20') #Tamaño de ventana y coordenadas en X e Y

# Creación de las etiquetas
mensaje = Label(root, text='Mi primer programa con Tkinter.')
mensaje2 = Label(root, text='Esta es la segunda etiqueta.')


# Se muestran las etiquetas
mensaje.grid(row=0,column=0) #Con el grid podemos colocar los elementos en diversas filas y columnas
mensaje2.grid(row=1,column=1) #Los grid también pueden aplicarse directamente en la creación de la etiqueta, ahorrando lineas de codigo

# Bucle para mantener la ventana en ejecución
root.mainloop()
````

### `pack` frente a `grid`: cuándo usar cada uno

En el archivo del día 08 se usó `pack()` porque solo había una etiqueta (da igual
la posición). En cuanto hay varios elementos que deben **alinearse entre sí**
(filas y columnas de un formulario), `grid()` es la herramienta correcta. El
proyecto del [día 12](#d012) lo aplica en un formulario real: las etiquetas van en la
columna 0 y los campos en la columna 1.

```python
Label(root, text='Nombre:').grid(row=0, column=0)
Entry(root).grid(row=0, column=1)
Label(root, text='Edad:').grid(row=1, column=0)
Entry(root).grid(row=1, column=1)
```

## Día 10 — Funciones lambda (anónimas)

**Archivo:** `d010.py`

---


### Qué se ve en este archivo

### Qué es una lambda

Una **lambda** es una función anónima: una función sin nombre, escrita en **una
sola línea** y con **una sola expresión**.

```python
def multiplicacion(numero1, numero2):
    return numero1 * numero2

multiplicacion = lambda numero1, numero2 : numero1 * numero2
```

Las dos versiones hacen **exactamente lo mismo**. La estructura es:

```python
lambda parámetros : expresión
```

No se puede escribir `return` (la expresión **es** el return implícito) ni
usar dos sentencias.

### Los dos usos del archivo

**1. Guardarla en una variable para reutilizarla**

```python
multiplicacion = lambda numero1, numero2 : numero1 * numero2
resultado1 = multiplicacion(10, 7)    # 70
resultado2 = multiplicacion(56, 5)    # 280
```

**2. Declararla y llamarla en la misma línea**

```python
(lambda numero1, numero2 : print(numero1 * numero2)) (7, 5)
```

Se lee: «crea esta función, y llámala con 7 y 5». Es la forma de **no darle
nombre**, útil cuando la función se usa una sola vez y no merece un `def`.

### El ejercicio del archivo

```python
PI = 3.14159265359
(lambda radio : print(f'El área del círculo es: {round(PI**radio,2)}'))(int(input('Introduce un radio\n')))
```

Se declara la lambda, se le pasa el radio leído por teclado, y el resultado se
imprime directamente sin guardar en ninguna variable.

````python
# FUNCIONES LAMBDA O ANONIMAS
# Solo pueden tener una expresión
# Se almacenan en una variable para poder llamarla, pero si solo se quiere usar una sola vez

multiplicacion = lambda numero1,numero2 : numero1*numero2
resultado1 = multiplicacion(10,7)
resultado2 = multiplicacion(56,5)

# declaracion y llamada en una misma linea
(lambda numero1,numero2 : print(numero1*numero2)) (7,5)

# EJERCICIO1: CREA CON UNA FUNCION LAMBDA UNA CALCULADORA DE ÁREAS DE CÍRCULO
PI = 3.14159265359
(lambda radio : print(f'El área del círculo es: {round(PI**radio,2)}') ) (int(input('Introduce un radio\n')))
````

### Aviso importante: el ejercicio del círculo tiene un error

Tal como está escrito, el programa imprime `PI ** radio`, que **no es el área del
círculo**. El área de un círculo es:

```python
area = PI * radio ** 2      # pi por el radio al cuadrado
```

El código del archivo calcularía otra cosa: con `radio = 2` imprimiría
`9.87` (que es pi²) en lugar de `12.57` (que es el área real). Es un buen recordatorio
de la diferencia entre `**` y `*`, y de que hay que comprobar que la fórmula
usada es la correcta.

La lambda corregida quedaría:

```python
PI = 3.14159265359
(lambda radio: print(f'El área del círculo es: {round(PI * radio ** 2, 2)}'))(int(input('Introduce un radio\n')))
```

### ¿Cuándo usar lambda y cuándo no?

| Situación | advisable |
| --- | --- |
| Función de una expresión, sin nombre, usada una vez | **Sí**, lambda |
| Varias líneas, con lógica o `if` | No, `def` |
| Función con nombre que se usa varias veces | No, `def` |
| Como argumento de `sorted()`, `map()`, `filter()` | **Sí**, lambda |

```python
# Lambda como argumento: aquí brilla
nombres = ['Ana', 'Zoe', 'Berta']
nombres.sort(key=lambda n: len(n))          # ordena por longitud
nums = [1, 2, 3, 4]
cuadrados = list(map(lambda n: n ** 2, nums))   # [1, 4, 9, 16]
```

### La alternativa con `def` + condición

Cuando una lambda empieza a necesitar `if`, ya no es una lambda: es una función
normal.

```python
# En vez de esto...
doble = lambda n: n * 2 if n > 0 else 0

# ...usa esto
def doble(n):
    return n * 2 if n > 0 else 0
```

### Resumen

- `lambda` = `def` de una línea, sin nombre, con una sola expresión.
- No admite `return` explícito ni varias sentencias.
- Se asigna a una variable si se va a reutilizar; si no, se invoca directamente.
- Prioriza la legibilidad: si cuesta leerla, conviene un `def`.

## Día 11 — Programación Orientada a Objetos (POO)

**Archivo:** `d011.py`

---


### Qué se ve en este archivo

### La idea

La POO sirve para agrupar **datos y las funciones que los usan** en una sola
entidad. Una **clase** es el molde; un **objeto** (o instancia) es una pieza
concreta hecha con ese molde.

Piensa en la diferencia entre el **plano** de una casa y una **casa concreta**:
el plano describe cómo es cualquier casa (atributos y métodos), y cada casa tiene
su propio color y su propio tamaño.

### La estructura de una clase

```python
class Vehiculo():                              # 1. Definir el molde

    # Atributos de CLASE: valores por defecto que comparten todos los objetos
    pais_origen = 'Alemania'

    # CONSTRUCTOR: se ejecuta al crear el objeto. Define los atributos de INSTANCIA
    def __init__(self, color, longitud, ruedas):
        self.color = color
        self.longitud = longitud
        self.ruedas = ruedas

    # MÉTODOS: funciones que pertenecen al objeto
    def arrancar(self):
        print('El vehículo ha arrancado.')

    def mostrar_info(self):
        print(f'Vehículo de color {self.color}...')

# Instancias: dos vehículos distintos
vehiculo_1 = Vehiculo('rojo', 4, 4)
vehiculo_2 = Vehiculo('negro', 8.25, 8)
vehiculo_1.mostrar_info()
```

### Los cuatro conceptos clave

1. **`class Vehiculo():`** — el molde. Los paréntesis están vacíos porque no
   hereda de nada (eso llega en temas posteriores).
2. **`__init__`** — el **constructor**. Se llama **solo** al crear el objeto
   (`= Vehiculo(...)`) y sirve para inicializar los atributos de esa instancia.
3. **`self`** — la referencia al propio objeto. Es lo que permite que cada
   vehículo tenga su propio color. **El primer parámetro de todo método es
   siempre `self`**, y se llama implícitamente: no hay que pasarlo.
   Por eso, al usar el método desde fuera, `mostrar_info()` se llama sin
   argumentos, pero dentro del método los atributos se escriben **siempre**
   `self.color`, nunca `color`.
4. **Atributos de clase vs de instancia:**

| Tipo | Dónde se declara | Compartido |
| --- | --- | --- |
| De clase | En el cuerpo de la clase, sin `self` | **Sí**, por todos los objetos |
| De instancia | En `__init__`, con `self.` | **No**, cada objeto tiene el suyo |

`pais_origen = 'Alemania'` es de clase: todos los vehículos son de Alemania,
y no hace falta pasarle ese dato al constructor. `self.color` es de instancia:
cada uno con su color.

### Clase vacía

```python
class Camion():
    pass
```

`pass` es una instrucción que **no hace nada**. Sirve para crear una clase vacía
(para añadirle cosas después) o como marcador de "aquí irá algo", que es una
práctica muy legible.

````python
#PROGRAMACIÓN ORIENTADA A OBJETOS (POO)

class Vehiculo(): 

    # Atributos de clase (valores por defecto que tienen todos)

    pais_origen = 'Alemania'


    def __init__(self, color, longitud,ruedas): #Constructor que tiene los atributos de instancia
        self.color = color
        self.longitud = longitud
        self.ruedas = ruedas

    # MÉTODOS
    def arrancar(self):
        print('El vehículo ha arrancado.')

    def detener(self):
        print('El vehículo se ha detenido.')

    # Cuando se llama a un atributo en un método siempre hay que ponerlo así:
    def mostrar_info(self):
        print(f'Vehículo de color {self.color}. Con una longitud de {self.longitud}. Tiene {self.ruedas} ruedas.')

# Instancia de objetos
vehiculo_1 = Vehiculo('rojo', 4, 4)
vehiculo_2 = Vehiculo('negro', 8.25, 8)

vehiculo_1.mostrar_info()

class Camion(): #Clase vacía 
    pass
````

### Flujo completo: crear un objeto y usarlo

```text
vehiculo_1 = Vehiculo('rojo', 4, 4)
                    │       │  │  │
                    │       └──┴──┴──┘  argumentos posicionales
                    └──────────────────  __init__(self, color, longitud, ruedas)
                                          self = vehiculo_1 (implícito)
                    self.color    = 'rojo'
                    self.longitud = 4
                    self.ruedas   = 4

vehiculo_1.mostrar_info()      ->  usa los atributos de vehiculo_1
vehiculo_2.mostrar_info()      ->  usa los de vehiculo_2
```

Los dos objetos son de la misma clase pero **completamente independientes**.
Cambiar `vehiculo_1.color` no afecta a `vehiculo_2`.

### Instanciar con argumentos por clave

```python
vehiculo_2 = Vehiculo(color='negro', longitud=8.25, ruedas=8)
```

Es la misma llamada que en el [día 06](#d006) se usaba con las funciones, y aquí
es aún más conveniente, porque las clases suelen tener muchos parámetros. El
proyecto [proyectod011](#proyectod011) usa las dos formas en el mismo archivo.

### El objeto `self`, en detalle

Dentro de un método, `self` **es** el objeto que se está usando. Estas dos
líneas hacen lo mismo:

```python
vehiculo_1.mostrar_info()      # lo que escribe el usuario
Vehiculo.mostrar_info(vehiculo_1)   # lo que ocurre por dentro
```

Por eso `self` es el primer parámetro de todo método, y por eso dentro hay que
escribir `self.color` y no `color`.

### Atributos que se crean sobre la marcha

En el proyecto de las motos, `consulta_precio()` usa `self.precio` sin que
`precio` esté en el constructor. Python lo permite: los atributos se crean en
cuanto se les asigna un valor (`motocicleta.precio = 27000`). Es una
particularidad de los lenguajes dinámicos, y conviene conocerla, aunque en un
proyecto ordenado es preferible declarar todo en `__init__`.

### Convenciones de nombres en POO

- **Clases** en `MayúsculaConPalabrasPegadas`: `Vehiculo`, `Motocicleta`.
- **Objetos, funciones y variables** en `minúsculas_con_guiones`: `vehiculo_1`.
- Los métodos que solo usan `self` se pueden declarar con el decorador
  `@staticmethod`, o con `@property` si queremos que se llamen como atributos.
- Un método `__str__` permite definir cómo se muestra el objeto al imprimirlo.

```text
     a. atributos de clase

```

Los apartados 2 y 3 se escriben **fuera** de la clase, con sangría de nivel 0.

## Día 12 — Tkinter: `Entry`, botones y eventos

**Archivo:** `d012.py`

---


### Qué se ve en este archivo

Este día une Tkinter con las funciones del [día 06](#d006): la ventana deja de
mostrar texto fijo, **recibe datos del usuario y reacciona a sus acciones**.

### 1. Campo de texto: `Entry`

```python
entrada = Entry(root)
entrada.insert(0, 'Escriba su nombre...')    # texto predefinido
entrada.bind('<Button-1>', lambda x: entrada.delete(0, END))   # se borra al hacer clic
entrada.pack()
```

- `Entry(root)` crea un campo vacío de una línea.
- `.insert(0, texto)` escribe dentro del campo. El `0` es la posición (al principio).
- `.bind(evento, función)` conecta una **acción del usuario** con una función.

### 2. El `lambda` como traductor de eventos

Este es el uso más importante del archivo y merece atención, porque no es un
truco: Tkinter **llama a la función con un argumento** (el objeto `evento`).

```python
entrada.bind('<Button-1>', lambda x: entrada.delete(0, END))
```

`x` es el evento. Es obligatorio escribirlo. Si se pusiera directamente
`entrada.delete(0, END)`, Tkinter ejecutaría el `delete` **al crear la ventana**,
no al hacer clic, y el texto desaparecería al instante. La lambda envuelve la
acción para que se ejecute **después**, con el evento como parámetro.

- `'<Button-1>'` = clic del botón izquierdo.
- `'<'` = al presionar. `'<ButtonRelease-1>'` = al soltar.
- `END` es una constante de Tkinter que significa «hasta el final». En
  `delete(0, END)` significa «borra todo el contenido».

### 3. Botones: `command`

```python
def pulsar_boton():
    nombre = entrada.get()                  # leer lo que escribió el usuario
    Label(root, text=f'{nombre}').pack()    # crear una etiqueta nueva

Button(root, text='Enviar', command=pulsar_boton).pack()
```

- `command` es la función que se ejecuta al pulsar. **Sin paréntesis**: si pones
  `command=pulsar_boton()`, la función se ejecutaría al crear el botón y la
  ventana se quedaría colgada.
- `.get()` **devuelve lo escrito** por el usuario. Sin `.get()`, `entrada` es solo
  el widget, no el texto.

### 4. `lambda` para pasar argumentos a un botón

Aquí el `command` llama a `crear_etiqueta(1)`, que **necesita un argumento**.
Como Tkinter no sabe qué argumento pasar, se usa una lambda que se encarga de
dárselo:

```python
def crear_etiqueta(numero_boton):
    etiqueta = Label(root, text=f'Botón {numero_boton} pulsado.')
    etiqueta.pack()

Button(root, text='Numero1', command=lambda: crear_etiqueta(1)).pack()
Button(root, text='Numero2', command=lambda: crear_etiqueta(2)).pack()
Button(root, text='Numero3', command=lambda: crear_etiqueta(3)).pack()
Button(root, text='Numero4', command=lambda: crear_etiqueta(4)).pack()
```

Los cuatro botones comparten la misma función y solo cambia el número. Es el
mismo patrón del `match` del [día 03](#d003): una lógica, muchos caminos de
entrada.

### Por qué las etiquetas nuevas no se ven

```python
Label(root, text=f'{nombre}').pack()
```

Cada vez que se pulsa el botón se **crea una etiqueta nueva** y se añade con
`pack()`. Como cada `pack()` apila, las etiquetas se van acumulando hacia abajo.

````python
# importaciones
from tkinter import * 

# Creación de la ventana principal
root = Tk()

# Creación de la etiqueta
mensaje = Label(root, text='Curso de Tkinter')

#Entrada de texto
entrada = Entry(root)
entrada.insert(0, 'Escriba su nombre...') #Sirve para que haya un texto predefinido en ese espacio
entrada.bind('<Button-1>', lambda x : entrada.delete(0, END)) # Sirve para que al pulsar con el ratón el texto predefinido se borre
# Necesito la función lambda para desencadenar el evento, porque si no lo borra instantáneamente
entrada.pack()

#Evento para el botón
def pulsar_boton():
    nombre = entrada.get()
    Label(root, text=f'{nombre}').pack()

def crear_etiqueta(numero_boton):
    etiqueta = Label(root, text=f'Botón {numero_boton} pulsado.')
    etiqueta.pack()

#Botón
Button(root, text='Enviar',command=pulsar_boton).pack() #La función pulsar_boton sin parentesis para un evento
Button(root, text='Numero1', command=lambda: crear_etiqueta(1)).pack()
Button(root, text='Numero2', command=lambda: crear_etiqueta(2)).pack()
Button(root, text='Numero3', command=lambda: crear_etiqueta(3)).pack()
Button(root, text='Numero4', command=lambda: crear_etiqueta(4)).pack()


# Bucle para mantener la ventana en ejecución
root.mainloop()
````

### Tabla de eventos usados en Tkinter

| Evento | Se dispara |
| --- | --- |
| `'<Button-1>'` | Clic del botón izquierdo |
| `'<Button-2>'` / `'<Button-3>'` | Clic central / derecho |
| `'<ButtonRelease-1>'` | Al soltar el botón |
| `'<Return>'` | Al pulsar ENTER |
| `'<Key>'` | Al pulsar cualquier tecla |
| `'<FocusIn>'` / `'<FocusOut>'` | Al entrar o salir del campo |

### Regla de oro de los callbacks

```text
command=funcion          ->  Tkinter no le pasa argumentos
command=funcion()        ->  se ejecuta AL CREAR el botón  (error típico)
command=lambda: f(1)     ->  se ejecuta al pulsar, pasando 1
command=lambda e: f(e)   ->  se ejecuta al pasar el evento
```

`command` es, en realidad, `bind('<Button-1>', ...)` por debajo: Tkinter siempre
intenta llamar a la función con el evento. Cuando la función no acepta
argumentos, la `lambda` sin parámetros (`lambda: ...`) hace de puente y descarta
el evento.

### El `Entry` es el `input` de las interfaces gráficas

| Consola | Gráfico |
| --- | --- |
| `input('Escribe: ')` | `entrada.get()` |
| `print(texto)` | `Label(root, text=texto)` |
| `if/else` sobre el dato | `if/else` dentro de la función del `command` |
| Espera a ENTER | Se dispara al pulsar el botón |

La lógica es **idéntica**: leer un dato, comprobarlo, mostrar un resultado. Lo
único que cambia es quién lo pide y cuándo.


Cómo funciona, paso a paso:

1. `fr.lower()` convierte el texto a minúsculas.
2. El `for` recorre **cada carácter** (incluidos espacios, comas y el salto de

3. `contador` empieza en 0 y se incrementa con `+= 1` en cada coincidencia.
4. La f-string presenta el resultado.

### Dos errores en este ejercicio

**1. La letra buscada no se convierte a minúsculas**

El texto sí se pasa a minúsculas, pero `t` no. Si el usuario escribe `A` en
mayúscula, `letra == t` nunca será cierto y el resultado será `0`. El arreglo:

```python
t = input('¿Qué letra quieres rastrear? \n').lower()
```

distingue nada: en un texto con `\n` al final, el bucle recorre ese carácter
también. Versión más robusta:

```python
for letra in fr.lower().replace(' ', ''):
```

Ambos casos tienen la misma moraleja: **normaliza los datos antes de
compararlos**. Es exactamente el mismo `.lower()` que se usa en `salida.lower()`

## Día 14 — Rutas de archivos con `os.path` en Tkinter

**Archivo:** `d014.py`

---


### Qué se ve en este archivo

### El problema que resuelve

Un programa no siempre encuentra sus archivos: depende de **desde dónde** se
ejecute. Si abres una terminal en `C:\` y lanzas `python D:\CursoPython\d014.py`,
la carpeta actual (`C:\`) **no** es la carpeta del programa. Por eso se
construye la ruta **a partir de la ubicación del propio archivo**.

### `__file__` y `os.path`

```python
import os                                    # módulo del sistema operativo
carpeta_principal = os.path.dirname(__file__) # la carpeta donde está el .py
carpeta_imagenes = os.path.join(carpeta_principal, 'imagenes')
```

| Pieza | Significado |
| --- | --- |
| `os` | Módulo que da acceso al sistema de archivos y al sistema operativo |
| `__file__` | Variable automática de Python con la **ruta del archivo actual** |
| `os.path.dirname(ruta)` | Devuelve **la carpeta** que contiene esa ruta |
| `os.path.join(a, b)` | **Une** dos partes de ruta con el separador correcto |

`__file__` es como el `self` de un archivo: una variable que el interprete
proporciona automáticamente.

### Separadores de ruta: por qué `join` y no `+`

Cada sistema operativo usa un separador distinto:

```text
Windows      ->  C:\CursoPython\imagenes
Linux/macOS  ->  /home/usuario/CursoPython/imagenes
```

`os.path.join()` pone el separador correcto automáticamente:

```python
os.path.join('C:\\CursoPython', 'imagenes')   # C:\CursoPython\imagenes
os.path.join('/home/user', 'imagenes')        # /home/user/imagenes
```

Unir con `+` o con `\` a mano **solo funciona en Windows** y rompe en cualquier
otro sistema. Es el error más común al empezar a manipular rutas.

### El icono de la ventana

```python
root.iconbitmap(os.path.join(carpeta_imagenes, 'python_icon.ico'))
```

`iconbitmap` cambia el icono de la barra de título. Necesita un archivo
`.ico` (o `.png` en Linux). En el proyecto está en la carpeta `imagenes\`, que es
justo donde se la apunta con `join`.

El programa termina con `root.mainloop()` sin ningún widget: la ventana aparece

```python
class Motocicleta():
```

### Los métodos

| Método | Qué hace |

| `detener()` | Apaga el motor, o avisa de que ya estaba apagado |

Los dos primeros son un ejemplo claro de **método que decide**: consultan el

`repostar()` usa un `while` para **no dejar que el usuario se pase de la

```python
litros_repostados = int(input('...'))
while (litros_repostados + self.combustible_litros) > self.capacidad:
    print('No cabe tanto combustible. ¿Quieres encharcar el concesionario?')
    litros_repostados = int(input('...'))
self.combustible_litros += litros_repostados
```

Es el bucle de validación del [día 05](#d005) aplicado a un atributo del objeto.

### Las dos formas de instanciar

```python
# 1. Con argumentos POSICIONALES
motocicleta_yamaha = Motocicleta('Roja y blanca', '45663-FHD', 10, 2, 'Yamaha',
                                 'YZF-R1', '22/02/2022', 288, 199, 17)

# 2. Con argumentos POR CLAVE (más legible con 10 parámetros)
motocicleta_harley = Motocicleta(matricula='48659-FHE', combustible_litros=0,
                                 color='Negra', marca='Harley Davidson',
                                 modelo='Fat Boy', numero_ruedas=2, peso=304,
                                 fecha_fabricacion='29/03/2021',
                                 velocidad_punta=160, capacidad=20)
```

Con diez parámetros, la segunda forma gana claramente: se lee sin tener que
contar posiciones. Es la misma idea que las funciones del [día 06](#d006), pero
con diez parámetros resulta todavía más necesaria.

````python
class Motocicleta():
    # Atributo de clase
    estado = 'nueva'
    motor = False

    #Constructor
    def __init__(self, color, matricula,combustible_litros, numero_ruedas,marca,modelo,fecha_fabricacion,velocidad_punta,peso,capacidad):
        self.color = color
        self.matricula = matricula
        self.combustible_litros = combustible_litros
        self.numero_ruedas = numero_ruedas
        self.marca = marca
        self.modelo = modelo
        self.fecha_fabricacion = fecha_fabricacion
        self.velocidad_punta = velocidad_punta
        self.peso = peso
        self.capacidad = capacidad
    
    def arrancar(self):
        if self.motor == False:
             self.motor = True
             print('El motor ha arrancado, ruge como un león.')
        else:
            print('Se escuha un molesto sonido al girar la llave, el motor ya estaba en marcha')

    def detener(self):
        if self.motor != False:
            self.motor = False
            print('El motor se ha detenido.')
        else:
            print('El motor ya estaba detenido.')

    def consulta_precio(self):
        print(f'El precio de la motocicleta {self.marca} {self.modelo} es de {self.precio}')

    def consultar_litros(self):
        print(f'--->REPORTE DE DEPÓSITO DE {self.marca} {self.modelo}<---')
        print(f'El depósito tiene {self.combustible_litros}.')
        print(f'La capacidad máxima del tanque de combustible es de {self.capacidad}')
        print(f'Faltan {self.capacidad - self.combustible_litros} para llenar el depósito.')
        print('--->FIN DEL REPORTE<---')
        
    def repostar(self):
        litros_repostados = int(input('Por favor, introduzca los litros que desea repostar:\n'))
        while (litros_repostados + self.combustible_litros) > self.capacidad:
            print('No cabe tanto combustible. ¿Quieres encharcar el concesionario?')
            litros_repostados = int(input('Por favor, introduzca los litros que desea repostar:\n'))
        self.combustible_litros+=litros_repostados
        print(f'Repostaje exitoso.\nSe han repostado {litros_repostados} litros.\nEl depósito tiene {self.combustible_litros} litros de combustible.')
       

# Instanciar un objeto

motocicleta_yamaha = Motocicleta('Roja y blanca', '45663-FHD', 10, 2, 'Yamaha','YZF-R1', '22/02/2022', 288, 199,17)
motocicleta_harley = Motocicleta(matricula = '48659-FHE', combustible_litros=0,color='Negra', marca='Harley Davidson',
modelo='Fat Boy', numero_ruedas=2,peso=304,fecha_fabricacion='29/03/2021', velocidad_punta=160, capacidad=20)

# Llamando a un método
motocicleta_yamaha.arrancar()
motocicleta_harley.detener()

# Asignar a uno de los dos objetos un atributo nuevo llamado precio
motocicleta_harley.precio = 27000
motocicleta_yamaha.precio = 21000
motocicleta_harley.consulta_precio()
motocicleta_yamaha.consulta_precio()

motocicleta_yamaha.consultar_litros()
motocicleta_yamaha.repostar()
````

### Detalle importante: el atributo de clase `motor`

`motor = False` está declarado en el cuerpo de la clase, como atributo **de
clase**. A primera vista parece un problema: ¿no es el mismo valor para todas las
motos?

La realidad es que **no**, por una sutileza de Python:

```text
arrancar():
    if self.motor == False:      # lee el atributo de clase: False
        self.motor = True        # ¡y aquí CREA un atributo de INSTANCIA!
```

Cuando dentro de un método se escribe `self.motor = True`, Python **no** modifica
el atributo de clase: crea un atributo propio **solo en ese objeto**. El resultado
es que cada moto tiene su propio estado del motor, y la clase conserva `False`
como valor por defecto. Es el **sombreado de atributos** (*attribute shadowing*).

La forma de evitar la confusión es declarar el atributo solo en `__init__`:

```python
def __init__(self, ...):
    self.motor = False           # atributo de instancia, sin sorpresas
```

### Detalle importante: `self.precio` no existe

`consulta_precio()` imprime `self.precio`, pero `precio` **no está** en el
constructor. Funciona porque antes se ha hecho:

```python
motocicleta_harley.precio = 27000
motocicleta_yamaha.precio = 21000
```

Python permite **crear atributos nuevos sobre un objeto ya creado**, sin previo
aviso. Es muy flexible, pero también es una fuente de errores: si se llama a
`consulta_precio()` sin haber asignado `precio`, aparece un `AttributeError` con
un mensaje poco claro.

La forma ordenada de evitarlo es declarar `precio` en `__init__` (con valor
`None` si aún no se conoce) o usar `getattr(self, 'precio', 'sin precio')`.

## Proyecto del día 12 — Formulario de nombre y edad (Tkinter)

**Archivo:** `proyectod012.py`

---


### Qué se ve en este archivo

Un formulario en Tkinter usando todo lo del [día 12](#d012) y el sistema de
colocación `grid` del [día 09](#d009).

### La estructura

```text
Fila 0:  [Nombre:]   [campo de texto]
Fila 1:  [Edad:]     [campo de texto]
Fila 2:              [ Enviar ]
Fila 3:              [ Tu nombre es X y tu edad es Y ]
```

Gráficamente:

```text
Nombre:  |_____________|
Edad:    |_____________|
         |   Enviar    |
```

### La construcción

```python
root = Tk()
root.title('Ejercicio con Tkinter')
root.geometry('300x200+560+50')

Label(root, text='Nombre:').grid(row=0, column=0)
Label(root, text='Edad:').grid(row=1, column=0)

entrada = Entry(root)
entrada.grid(row=0, column=1)
entrada1 = Entry(root)
entrada1.grid(row=1, column=1)

def pulsar_boton():
    nombre = entrada.get()
    edad = entrada1.get()
    Label(root, text=f'Tu nombre es {nombre} y tu edad es {edad}').grid(row=3, column=1)

Button(root, text='Enviar', command=pulsar_boton).grid(row=2, column=1)

root.mainloop()
```

Piezas clave:

- **`grid` con dos columnas**: etiquetas en la `column=0`, campos en la
  `column=1`. Esto es lo que alinea el formulario. La posición de la etiqueta y
  la del campo es la misma fila (`row=0`), columnas distintas.
- **Los widgets se guardan en variables** (`entrada`, `entrada1`) porque hay que
  llamar a `.get()` sobre ellos después.
- **La etiqueta del resultado se crea dentro de la función**, en la `row=3`. Al
  pulsar varias veces se apilan en la misma celda, así que se ven una debajo de
  otra.
- `command=pulsar_boton` **sin paréntesis**: así el botón no ejecuta nada al
  crearse.

### El comentario sobre el `get()`

En el archivo hay una nota: *"el grid no va seguido porque si no da error de None
en el get"*. Es un detalle práctico de Tkinter: si se escribe

```python
entrada = Entry(root).grid(row=0, column=1)
```

el valor de `entrada` es el **resultado de `grid()`**, que en Python devuelve
`None`. `entrada.get()` falla entonces con
`AttributeError: 'NoneType' object has no attribute 'get'`.

La regla es sencilla: **si necesitas el widget, sepáralo en dos líneas.**

```python
# Bien
entrada = Entry(root)
entrada.grid(row=0, column=1)

# Mal si luego vas a usar entrada
entrada = Entry(root).grid(row=0, column=1)
```

En Tkinter, casi todos los métodos de colocación devuelven `None`. La
abreviatura del [día 09](#d009) es válida solo para widgets que no vas a
manipular después.

````python
# importaciones
from tkinter import * # Del módulo tkinter importalo todo

# Creación de la ventana principal
root = Tk()

# Título de la ventana
root.title('Ejercicio con Tkinter')

# Tamaño de la ventana
root.geometry('300x200+560+50') #Tamaño de ventana y coordenadas en X e Y

# Creación de las etiquetas
Label(root, text='Nombre:').grid(row=0,column=0)
Label(root, text='Edad:').grid(row=1,column=0)
#Creación de entradas
entrada = Entry(root) # el grid no va seguido porque si no da error de None en el get
entrada.grid(row=0,column=1)
entrada1 = Entry(root)
entrada1.grid(row=1,column=1)

# Evento para el botón
def pulsar_boton():
    nombre = entrada.get()
    edad = entrada1.get()
    Label(root, text=f'Tu nombre es {nombre} y tu edad es {edad}').grid(row=3, column=1)

# Creación del botón
Button(root, text='Enviar',command=pulsar_boton).grid(row=2, column=1)


# Bucle para mantener la ventana en ejecución
root.mainloop()
````

### El recorrido completo de un clic

```text
1. El usuario hace clic en 'Enviar'
2. Tkinter llama a pulsar_boton()            <- command, sin paréntesis
3. entrada.get()  ->  'Ana'                <- lee el campo
4. entrada1.get() ->  '30'
5. Se crea un Label con el texto combinado
6. .grid(row=3, column=1) la coloca debajo
7. mainloop() sigue esperando más clics
```

### El formulario sin validar, y cómo mejorarlo

Fíjate en que la función **no valida** los datos: si el usuario no escribe nada,
el programa imprime «Tu nombre es y tu edad es». Es un detalle a mejorar, y la
solución usa el condicional del [día 03](#d003):

```python
def pulsar_boton():
    nombre = entrada.get().strip()
    edad = entrada1.get().strip()

    if not nombre or not edad:
        Label(root, text='Rellena los dos campos.').grid(row=3, column=1)
    elif not edad.isdigit():
        Label(root, text='La edad debe ser un número.').grid(row=3, column=1)
    else:
        Label(root, text=f'Tu nombre es {nombre} y tu edad es {edad}').grid(row=3, column=1)
```

- `.strip()` quita espacios sobrantes.
- `if not nombre` aprovecha que la cadena vacía es "falsy" en Python.
- `edad.isdigit()` es el método que comprueba si un texto son solo dígitos: la
  alternativa sería `int(edad)` dentro de un `try/except`.

### Cómo mejorarlo

- **Mostrar el resultado en la misma etiqueta** en vez de crear una nueva cada
  vez, guardándola en una variable:
  `etiqueta_resultado = Label(root, text='')` y luego
  `etiqueta_resultado.config(text=...)`.
- **Limpiar los campos** tras enviar, con `.delete(0, END)`.
- **Pulsar ENTER** para enviar:
  `root.bind('<Return>', lambda e: pulsar_boton())`.
- **Convertir la edad a número** y hacer cálculos con ella.
- Organizar todo en una **clase** `Formulario` con `__init__` y `crear_widgets`,
  aplicando la POO del [día 11](#d011).
